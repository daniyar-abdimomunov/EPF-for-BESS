# Example Notebook 05:

Run experiments with different configurations.

```python
%load_ext autoreload
%autoreload 2

import mlflow

from src import execute_run
import config
import itertools
```

```python
# Instantiate
seeds = [
    123,
    456,
    7891011,
    12131415,
    643216842,
]
datasets = [
    config.DATA_DE_CONFIG,
    config.DATA_DK_CONFIG,
    config.DATA_ES_CONFIG,
]
penalties = [
    config.COV_E_PENALTY_CONFIG,
    config.CORR_F_PENALTY_CONFIG,
    config.SPOPLUS_PENALTY_CONFIG,
]
```

```python
# Set tracking URI directly to local MLFlow database.
# mlflow.set_tracking_uri("sqlite:///../mlflow.db")
# ... or, set tracking URI to default local port when using MLFlow UI. Run `mlflow ui` in terminal to initialize MLFlow UI.
mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
# Pre-train models for fine-tuning
RUN_CONFIG = config.PRETRAIN_RUN_CONFIG

for seed, dataset in itertools.product(seeds, datasets):
    execute_run(
        **config.EXP_CONFIG,
        **RUN_CONFIG,
        data_config = dataset,
        seed = seed,
        tracking_uri = mlflow.get_tracking_uri(),
    )
```

```python
# Fine-tune pre-trained models
RUN_CONFIG = config.FINETUNE_RUN_CONFIG

for seed, dataset, penalty in itertools.product(seeds, datasets, penalties):
    execute_run(
        **config.EXP_CONFIG,
        **RUN_CONFIG,
        data_config = dataset,
        penalty_config = penalty,
        seed = seed,
        tracking_uri = mlflow.get_tracking_uri(),
    )
```

```python
# Train Composite Loss directly
RUN_CONFIG = config.TRAIN_RUN_CONFIG

for seed, dataset, penalty in itertools.product(seeds, datasets, [None] + penalties):
    execute_run(
        **config.EXP_CONFIG,
        **RUN_CONFIG,
        data_config = dataset,
        penalty_config = penalty,
        seed = seed,
        tracking_uri = mlflow.get_tracking_uri(),
    )
```
