# Example Notebook 01:
Optimization Model to solve for optimal BESS charging-discharging schedule

```python
%load_ext autoreload
%autoreload 2

import pandas as pd

from src import BESSSchedulingOptModel
from src.utils import plot_storage_schedule, solve_storage_schedule
import config
```

```python
# Read data into DataFrame.
de = pd.read_csv('../timexer/dataset/EPF/DE.csv', parse_dates=['date'])
```

```python
# Prepare prices for specific date.
prices = de['OT']
prices.index = de['date']
prices = prices['2017-12-22':'2017-12-29']
prices
```

```python
# Set up optimization model.
num_timesteps = len(prices)
model = BESSSchedulingOptModel(**config.BATTERY_CONFIG, H=num_timesteps)
```

```python
# Solve for optimal BESS charging-discharging schedule.
profit, charge_sol, discharge_sol, soc = solve_storage_schedule(prices, model)
print(f'Profit: {profit:.2f} EUR')
```

```python
# Plot charging-discharging schedule.
plot_storage_schedule(
    charge_sol,
    discharge_sol,
    prices.to_numpy(),
    soc,
)
```
